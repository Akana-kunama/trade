```python
import pandas as pd
from pathlib import Path

def _build_trade_calendar(parquet_files, start=None, end=None):
    """
    用所有ETF的index并集构造“交易日历”（无需外部交易日历接口）
    """
    all_dates = None
    start = pd.to_datetime(start) if start is not None else None
    end = pd.to_datetime(end) if end is not None else None

    for fp in parquet_files:
        df = pd.read_parquet(fp)
        idx = pd.to_datetime(df.index)

        if start is not None:
            idx = idx[idx >= start]
        if end is not None:
            idx = idx[idx <= end]

        all_dates = idx if all_dates is None else all_dates.union(idx)

    if all_dates is None:
        return pd.DatetimeIndex([])
    return pd.DatetimeIndex(all_dates.sort_values().unique())

def _etf_metrics(df, calendar, start, end):
    """
    计算一个ETF在区间内的“可交易完整性指标”
    """
    start = pd.to_datetime(start)
    end = pd.to_datetime(end)

    df = df.copy()
    df.index = pd.to_datetime(df.index)
    df = df.sort_index()

    # 对齐到统一交易日历
    cal = calendar[(calendar >= start) & (calendar <= end)]
    aligned = df.reindex(cal)

    # 定义“有效交易日”：close 非空 且 amount>0
    valid = aligned["close"].notna() & aligned["amount"].fillna(0).gt(0)

    # first_valid_date
    first_valid_date = valid.idxmax() if valid.any() else pd.NaT

    # coverage
    coverage = float(valid.mean()) if len(valid) else 0.0

    # max consecutive missing
    missing = ~valid
    max_gap = 0
    cur = 0
    for m in missing.values:
        if m:
            cur += 1
            if cur > max_gap:
                max_gap = cur
        else:
            cur = 0

    return {
        "first_valid_date": first_valid_date,
        "coverage": coverage,
        "max_consecutive_missing": int(max_gap),
        "n_days": int(len(valid)),
        "n_valid": int(valid.sum()),
    }

def select_etf_universe(
    parquet_dir,
    start,
    end,
    min_coverage=0.98,
    max_consecutive_missing=10,
    start_grace_days=5,
    code_col="code",
):
    """
    输入：目录 + 时间段 + 条件
    输出：符合要求的ETF code 列表 + 指标表(df_metrics)
    """
    parquet_dir = Path(parquet_dir)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        raise FileNotFoundError(f"No parquet files found in: {parquet_dir}")

    start = pd.to_datetime(start)
    end = pd.to_datetime(end)

    # 构造本地交易日历
    calendar = _build_trade_calendar(files, start=start, end=end)

    rows = []
    for fp in files:
        df = pd.read_parquet(fp)

        # code：优先从列里取，否则从文件名推断
        if code_col in df.columns and df[code_col].notna().any():
            code = str(df[code_col].dropna().iloc[0])
        else:
            # 兼容 510050.SH.parquet 或 510050_SH.parquet
            stem = fp.stem
            code = stem.replace("_", ".")  # safe_name 反推

        m = _etf_metrics(df, calendar, start, end)
        rows.append({"code": code, "file": str(fp), **m})

    df_metrics = pd.DataFrame(rows)

    # 通过条件
    start_grace_limit = start + pd.Timedelta(days=start_grace_days)
    passed_mask = (
        df_metrics["first_valid_date"].notna()
        & (df_metrics["first_valid_date"] <= start_grace_limit)
        & (df_metrics["coverage"] >= min_coverage)
        & (df_metrics["max_consecutive_missing"] <= max_consecutive_missing)
    )

    df_pass = df_metrics[passed_mask].sort_values(
        ["coverage", "first_valid_date"], ascending=[False, True]
    ).reset_index(drop=True)

    df_fail = df_metrics[~passed_mask].sort_values(
        ["coverage", "first_valid_date"], ascending=[True, False]
    ).reset_index(drop=True)

    codes = df_pass["code"].tolist()
    return codes, df_pass, df_fail
```
